```python
## If we have the header
df = spark.read\
    .option("header", "true")\
    .option("inferSchema", "true")\
    .csv("/Volumes/databeli/test_tables/csv_files/raw_csv_files/sales.csv")
display(df)
```

```python
## If we have the no header
schema = "id integer, customer_name string, product string, quantity integer, price integer, remarks string"

df = spark.read\
    .option("header", "true")\
    .schema(schema)\
    .option("inferSchema", "true")\
    .csv("/Volumes/databeli/test_tables/csv_files/raw_csv_files/sales_no_header.csv")
display(df)
```

```python
## Or we can define schema with another approach
from pyspark.sql.types import StructType, StructField, IntegerType, StringType, DoubleType

schema = StructType([
    StructField("id", IntegerType(), True),
    StructField("customer_name", StringType(), True),
    StructField("product", StringType(), True),
    StructField("quantity", IntegerType(), True),
    StructField("price", IntegerType(), True),
    StructField("remarks", StringType(), True)]
)

df = spark.read\
    .option("header", "true")\
    .schema(schema)\
    .option("inferSchema", "true")\
    .csv("/Volumes/databeli/test_tables/csv_files/raw_csv_files/sales_no_header.csv")
display(df)
```

```python
## If we have any other separator
df = spark.read\
    .option("header", "true")\
    .option("inferSchema", "true")\
    .option("delimiter","|")\
    .csv("/Volumes/databeli/test_tables/csv_files/raw_csv_files/sales_pipe.csv")
display(df)
```

```python
## If we have quote character
df = spark.read\
    .option("header", "true")\
    .option("inferSchema", "true")\
    .option("delimiter","|")\
    .option("quote","'")\
    .csv("/Volumes/databeli/test_tables/csv_files/raw_csv_files/sales_pipe_single_quote.csv")
display(df)
```

```python
## If we have escape character
df = spark.read\
    .option("header", "true")\
    .option("inferSchema", "true")\
    .option("delimiter","|")\
    .option("quote","'")\
    .option("escape","#")\
    .csv("/Volumes/databeli/test_tables/csv_files/raw_csv_files/sales_pipe_single_quote_escape.csv")
display(df)
```

```python
## If we have multi line text
df = spark.read\
    .option("header", "true")\
    .option("inferSchema", "true")\
    .option("delimiter","|")\
    .option("quote","'")\
    .option("escape","#")\
    .option("multiline", "true")\
    .csv("/Volumes/databeli/test_tables/csv_files/raw_csv_files/sales_pipe_single_quote_escape_multiline.csv")
display(df)
```
