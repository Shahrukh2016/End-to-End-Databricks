Working with JSON

```python
df = spark.read.format("json").load("/Volumes/databeli/test_tables/json_files/data_corrupted.json")
display(df)
```

```python
df = spark.read\
    .format("json")\
        .option("mode", "DROPMALFORMED")\
            .load("/Volumes/databeli/test_tables/json_files/data_corrupted.json")
display(df)
```

```python
df = spark.read\
    .format("json")\
        .option("mode", "FAILFAST")\
            .load("/Volumes/databeli/test_tables/json_files/data_corrupted.json")
display(df)
```
