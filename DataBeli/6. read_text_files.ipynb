```python
## If we want whole file as a single row
df = spark.read.format("text").load("/Volumes/databeli/test_tables/unstructured_files/log.txt",wholeText=True)
display(df)
```

```python
## If the data is completely unstructured like photos/videos then use binaryFile option for it
df = spark.read.format("binaryFile").load("/Volumes/databeli/test_tables/unstructured_files/log.txt",wholeText=True)
display(df)
```

```python
## If we want to load metadata as well
df = spark.read.format("text").load("/Volumes/databeli/test_tables/unstructured_files/log.txt",wholeText=True)
df2 = df.selectExpr("value", "_metadata.file_path", "_metadata.file_name", "_metadata.file_size", "_metadata.file_modification_time")
display(df2)
```
