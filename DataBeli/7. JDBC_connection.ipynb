```python
## creating JDBC connection
jdbc_url = "your jdbc url"

access_token = "your access token"

## SQL query or table
query = "(SELECT * FROM samples.bakehouse.sales_franchises LIMIT 100) as t"

## Redd data
df = spark.read\
    .format("jdbc")\
        .option("url", jdbc_url)\
            .option("query", query)\
                .option("user", "token")\
                    .option("password", access_token)\
                        .load()

display(df)
```

```python
## OR we can directly query the table

## creating JDBC connection
jdbc_url = "your jdbc url"

access_token = "your access token"

## Redd data
df = spark.read\
    .format("jdbc")\
        .option("url", jdbc_url)\
            .option("dbtable", "samples.bakehouse.sales_franchises")\
                .option("user", "token")\
                    .option("password", access_token)\
                        .load()

display(df)
```
