```python
from pyspark.sql.types import StructType, StructField, StringType, DoubleType, IntegerType, DateType
from datetime import date, timedelta
import random

schema = StructType([
    StructField("date", DateType(), True),
    StructField("location", StringType(), True),
    StructField("temperature_c", DoubleType(), True),
    StructField("humidity_percent", IntegerType(), True),
    StructField("wind_speed_kmh", DoubleType(), True)
])

locations = ["New York", "Los Angeles", "Chicago", "Houston", "Phoenix"]
start_date = date(2025, 11, 1)
num_days = 30

data = []
for i in range(num_days):
    for loc in locations:
        data.append((
            start_date + timedelta(days=i),
            loc,
            round(random.uniform(-5, 35), 1),
            random.randint(20, 100),
            round(random.uniform(0, 40), 1)
        ))

df = spark.createDataFrame(data, schema)
display(df)
```

```python
df.createOrReplaceTempView("weather")  ## session scoped
df.createOrReplaceGlobalTempView("global_weather")  ## globally scoped
```

```python
%sql
select * from weather;
```
