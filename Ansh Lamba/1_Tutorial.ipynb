### 2. Schema Definitions

#### 1. DDL Schema 

```python
## Checking the schema
df.printSchema()
```

```python
## Updating datatype of Item_Weight from double to string
my_ddl_schema = """
                    Item_Identifier STRING,
                    Item_Weight STRING,
                    Item_Fat_Content STRING, 
                    Item_Visibility DOUBLE,
                    Item_Type STRING,
                    Item_MRP DOUBLE,
                    Outlet_Identifier STRING,
                    Outlet_Establishment_Year INT,
                    Outlet_Size STRING,
                    Outlet_Location_Type STRING, 
                    Outlet_Type STRING,
                    Item_Outlet_Sales DOUBLE
"""
```

```python
## Loading data
df_schema = (
    spark.read.format("csv")
    .option("header", "true")
    .schema(my_ddl_schema)
    .table("ansh_lamba.default.big_mart_sales")
)
```

```python
## Checking data
df_schema.display()
```

```python
df_schema.printSchema()
```

#### 2. StructType() Schema 

```python
## Importing libraries
from pyspark.sql.functions import *
from pyspark.sql.types import *
```

```python
## defining schema
my_struct_schema = StructType(
    [
        StructField("Item_Identifier", StringType(), True),
        StructField("Item_Weight", StringType(), True),
        StructField("Item_Fat_Content", StringType(), True),
        StructField("Item_Visibility", StringType(), True),
        StructField("Item_MRP", StringType(), True),
        StructField("Outlet_Identifier", StringType(), True),
        StructField("Outlet_Establishment_Year", StringType(), True),
        StructField("Outlet_Size", StringType(), True),
        StructField("Outlet_Location_Type", StringType(), True),
        StructField("Outlet_Type", StringType(), True),
        StructField("Item_Outlet_Sales", StringType(), True),
    ]
)
```

```python
## Loading data
df_struct_schema = (
    spark.read.format("csv")
    .option("header", "true")
    .schema(my_struct_schema)
    .table("ansh_lamba.default.big_mart_sales")
)
```

```python
## display table
df_struct_schema.display()
```

### 3. Basic Functions

```python
df.limit(5).display()
```

```python
## Select few columns
df.select('Item_Identifier', 'Item_Weight', 'Item_Fat_Content').display()
```

```python
## Select few columns
df.select(col('Item_Identifier'), col('Item_Weight'), col('Item_Fat_Content')).display()
```

```python
## Aliasing
from pyspark.sql.functions import col

df.select(col('Item_Identifier').alias('Item_ID')).display()
```

```python
## Easy Slicing - Scenario 1
df.filter(col('Item_Fat_Content') == 'Regular').display()
```

```python
## Medium Slicing - Scenario 2
df.filter((col('Item_Weight') < 10) & (col('Item_Type') == 'Soft Drinks')).display()
```

```python
## Hard Slicing - Scenario 3
df.filter((col('Outlet_Location_Type').isin('Tier 1', 'Tier 2')) & (col('Outlet_Size').isNull())).display()
```

```python
## withColumnRenamed
df.withColumnRenamed('Item_Weight', 'Item_Wt').display()
```

```python
## withColumn - New Column - Scenario 1
from pyspark.sql.functions import *
df.withColumn('flag', lit('New')).display()
```

```python
## withColumn - New Column - Scenario 2
from pyspark.sql.functions import *
df.withColumn('multiply', round((col('Item_Weight')*col('Item_MRP')),2)).display()
```

```python
## withColumn - Updating Existing Column - Scenario 3
from pyspark.sql.functions import *
df.withColumn('Item_Fat_Content', regexp_replace(col('Item_Fat_Content'), 'Regular', 'Reg'))\
    .withColumn('Item_Fat_Content', regexp_replace(col('Item_Fat_Content'), 'Low Fat', 'LF'))\
        .display()
```

```python
## Typecasting
df.withColumn('Item_Weight', col('Item_Weight').cast('string')).display()
```

```python
## Sort - Scenario - 1
df.sort(col('Item_Weight').desc()).display()
```

```python
## Sort - Scenario - 2
df.sort([col('Item_Weight'), col('Item_Visibility')], ascending = [True, False]).display()
```

```python
## Limit
df.limit(5).display()
```

```python
## Drop - Scenario 1: Single Column Drop
df.drop(col('Item_Visibility')).display()
```

```python
## Drop - Scenario 2: Multiple Column Drop
df.drop(col('Item_Visibility'), col('Item_Type')).display()
```

```python
## Drop Duplicates: Scenario - 1 
df.dropDuplicates().display()
```

```python
## Drop Duplicates: Scenario - 2
df.dropDuplicates(['Item_Type']).display()
```

```python
## Distinct
df.distinct().display()
```

```python
## Preparing dataframes for further operations
data1 = [('1','kad'),
        ('2','sid')]
schema1 = 'id STRING, name STRING' 

df1 = spark.createDataFrame(data1,schema1)

data2 = [('3','rahul'),
        ('4','jas')]
schema2 = 'id STRING, name STRING' 

df2 = spark.createDataFrame(data2,schema2)
df1.display(), df2.display()

```

```python
## UNION
df1.union(df2).display()
```

```python
## Preparing data for UNION BY NAME
data1 = [('kad', '1'),
        ('sid','2')]
schema1 = 'name STRING, id STRING' 

df1 = spark.createDataFrame(data1,schema1)

data2 = [('3','rahul'),
        ('4','jas')]
schema2 = 'id STRING, name STRING' 

df2 = spark.createDataFrame(data2,schema2)
df1.display(), df2.display()
```

```python
## UNION BY NAME
df1.unionByName(df2).display()
```

```python
#### String Functions:
## Initcap():
df.select(initcap(col('Item_Type'))).display()
```

```python
#### String Functions:
## lower():
df.select(lower(col('Item_Type'))).display()
```

```python
#### String Functions:
## upper():
df.select(upper(col('Item_Type')).alias('upper_item_type')).display()
```

```python
#### Date Functions:
## Current Date:
df = df.withColumn('current_date', current_date())
df.display()
```

```python
#### Date Functions:
## Date Add:
df = df.withColumn('week_after', date_add(current_date(), 7))
df.display()
```

```python
#### Date Functions:
## Date Sub:
df = df.withColumn('week_before', date_sub(current_date(), 7))
df.display()
```

```python
#### Date Functions:
## Date Diff:
df = df.withColumn('date_diff', datediff(col('week_after'), col('current_date')))
df.display()
```

```python
#### Date Functions:
## Date Format:
df = df.withColumn('week_before', date_format(col('week_before'), 'dd-MM-yyyy'))
df.display()
```

```python
#### Handling Nulls:
## Dropting Nulls:
df.dropna('all').display()
```

```python
#### Handling Nulls:
## Dropting Nulls:
df.dropna('any').display()
```

```python
#### Handling Nulls:
## Dropting Nulls:
df.dropna(subset = ['Outlet_Size']).display()
```

```python
#### Handling Nulls:
## Filling Nulls:
df.fillna({'Item_Weight': 0, 'Outlet_Size': 'NA'}).display()
```

```python
#### Split and Indexing:
## Split:
df.withColumn('Outlet_Type', split(col('Outlet_Type'), ' ')).display()
```

```python
#### Split and Indexing:
## Split and Indexing:
df.withColumn('Outlet_Type', split(col('Outlet_Type'), ' ')).withColumn('Tier', element_at(col('Outlet_Type'), 2)).display()
```

```python
#### Explode:
## Preparing data for exploding
df_exp = df.withColumn('Outlet_Type', split(col('Outlet_Type'), ' '))
df_exp.display()
```

```python
#### Explode:
df_exp.withColumn('Outlet_Type', explode(col('Outlet_Type'))).display()
```

```python
#### ARRAY_CONTAINS
df.withColumn('Outlet_Type', split(col('Outlet_Type'), ' ')).withColumn('contains', array_contains(col('Outlet_Type'), 'Supermarket')).display()
```

```python
#### GroupBy
## Scenario 1
df.groupBy(col('Item_Type')).agg({'Item_MRP' : 'sum'}).display()
```

```python
#### GroupBy
## Scenario 2 - Grouping on 2 or more columns
df.groupBy(col('Item_Type') , col('Outlet_Size')).agg(sum('Item_MRP').alias('sum'), avg('Item_MRP').alias('avg')).orderBy(col('Item_Type')).display()
```
