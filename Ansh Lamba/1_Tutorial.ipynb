### 1. Data Reading 

#### 1. JSON Data Reading 

```python
## Loading data
df1 = spark.read.format('json').options(header='true', inferSchema='true')\
    .options(multiline='true')\
    .table('ansh_lamba.default.drivers')
```

```python
## Displaying data
df1.display()
```

#### 2. CSV Data Reading 

```python
## Loading data
df = spark.read.format('csv').options(header='true', inferSchema='true').table('ansh_lamba.default.big_mart_sales')
```

```python
## Displaying data
df.display()
```

### 2. Schema Definitions

#### 1. DDL Schema 

```python
## Checking the schema
df.printSchema()
```

```python
## Updating datatype of Item_Weight from double to string
my_ddl_schema = '''
                    Item_Identifier STRING,
                    Item_Weight STRING,
                    Item_Fat_Content STRING, 
                    Item_Visibility DOUBLE,
                    Item_Type STRING,
                    Item_MRP DOUBLE,
                    Outlet_Identifier STRING,
                    Outlet_Establishment_Year INT,
                    Outlet_Size STRING,
                    Outlet_Location_Type STRING, 
                    Outlet_Type STRING,
                    Item_Outlet_Sales DOUBLE
'''
```

```python
## Loading data
df_schema = spark.read.format('csv')\
    .option('header', 'true')\
    .schema(my_ddl_schema)\
    .table('ansh_lamba.default.big_mart_sales')
```

```python
## Checking data
df_schema.display()
```

```python
df_schema.printSchema()
```
