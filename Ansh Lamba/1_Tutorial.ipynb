### 2. Schema Definitions

#### 1. DDL Schema 

```python
## Checking the schema
df.printSchema()
```

```python
## Updating datatype of Item_Weight from double to string
my_ddl_schema = """
                    Item_Identifier STRING,
                    Item_Weight STRING,
                    Item_Fat_Content STRING, 
                    Item_Visibility DOUBLE,
                    Item_Type STRING,
                    Item_MRP DOUBLE,
                    Outlet_Identifier STRING,
                    Outlet_Establishment_Year INT,
                    Outlet_Size STRING,
                    Outlet_Location_Type STRING, 
                    Outlet_Type STRING,
                    Item_Outlet_Sales DOUBLE
"""
```

```python
## Loading data
df_schema = (
    spark.read.format("csv")
    .option("header", "true")
    .schema(my_ddl_schema)
    .table("ansh_lamba.default.big_mart_sales")
)
```

```python
## Checking data
df_schema.display()
```

```python
df_schema.printSchema()
```

#### 2. StructType() Schema 

```python
## Importing libraries
from pyspark.sql.functions import *
from pyspark.sql.types import *
```

```python
## defining schema
my_struct_schema = StructType(
    [
        StructField("Item_Identifier", StringType(), True),
        StructField("Item_Weight", StringType(), True),
        StructField("Item_Fat_Content", StringType(), True),
        StructField("Item_Visibility", StringType(), True),
        StructField("Item_MRP", StringType(), True),
        StructField("Outlet_Identifier", StringType(), True),
        StructField("Outlet_Establishment_Year", StringType(), True),
        StructField("Outlet_Size", StringType(), True),
        StructField("Outlet_Location_Type", StringType(), True),
        StructField("Outlet_Type", StringType(), True),
        StructField("Item_Outlet_Sales", StringType(), True),
    ]
)
```

```python
## Loading data
df_struct_schema = (
    spark.read.format("csv")
    .option("header", "true")
    .schema(my_struct_schema)
    .table("ansh_lamba.default.big_mart_sales")
)
```

```python
## display table
df_struct_schema.display()
```

### 3. Basic Functions

```python
df.limit(5).display()
```

```python
## Select few columns
df.select('Item_Identifier', 'Item_Weight', 'Item_Fat_Content').display()
```

```python
## Select few columns
df.select(col('Item_Identifier'), col('Item_Weight'), col('Item_Fat_Content')).display()
```

```python
## Aliasing
from pyspark.sql.functions import col

df.select(col('Item_Identifier').alias('Item_ID')).display()
```

```python
## Easy Slicing - Scenario 1
df.filter(col('Item_Fat_Content') == 'Regular').display()
```

```python
## Medium Slicing - Scenario 2
df.filter((col('Item_Weight') < 10) & (col('Item_Type') == 'Soft Drinks')).display()
```

```python
## Hard Slicing - Scenario 3
df.filter((col('Outlet_Location_Type').isin('Tier 1', 'Tier 2')) & (col('Outlet_Size').is3##Null())).display()
```

```python
## withColumnRenamed
df.withColumnRenamed('Item_Weight', 'Item_Wt').display()
```

```python
## withColumn - New Column - Scenario 1
from pyspark.sql.functions import *
df.withColumn('flag', lit('New')).display()
```

```python
## withColumn - New Column - Scenario 2
from pyspark.sql.functions import *
df.withColumn('multiply', round((col('Item_Weight')*col('Item_MRP')),2)).display()
```

```python
## withColumn - Updating Existing Column - Scenario 3
from pyspark.sql.functions import *
df.withColumn('Item_Fat_Content', regexp_replace(col('Item_Fat_Content'), 'Regular', 'Reg'))\
    .withColumn('Item_Fat_Content', regexp_replace(col('Item_Fat_Content'), 'Low Fat', 'LF'))\
        .display()
```

```python
## Typecasting
df.withColumn('Item_Weight', col('Item_Weight').cast('string')).display()
```

```python
## Sort - Scenario - 1
df.sort(col('Item_Weight').desc()).display()
```

```python
## Sort - Scenario - 2
df.sort([col('Item_Weight'), col('Item_Visibility')], ascending = [True, False]).display()
```

```python
## Limit
df.limit(5).display()
```

```python
## Drop - Scenario 1: Single Column Drop
df.drop(col('Item_Visibility')).display()
```

```python
## Drop - Scenario 2: Multiple Column Drop
df.drop(col('Item_Visibility'), col('Item_Type')).display()
```

```python
## Drop Duplicates: Scenario - 1 
df.dropDuplicates().display()
```

```python
## Drop Duplicates: Scenario - 2
df.dropDuplicates(['Item_Type']).display()
```
