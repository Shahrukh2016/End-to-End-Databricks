```python
df = spark.table('workspace.default.imdb_movies')
display(df.limit(10))
```

```python
%sql
select distinct studio from workspace.default.imdb_movies;
```

```python
rep_by_key = df.repartition(6, 'studio')
rep_by_key.explain('formatted')
```

```python
rep_by_rr = df.repartition(6)
rep_by_rr.explain('formatted')
```

```python
out_path = '/Volumes/workspace/default/repartition_demo/repartition_6'
rep_by_key.write.mode('overwrite').parquet(out_path)
```
